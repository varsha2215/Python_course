```python
# type of arguments
#1.positional arguments: the arguments are assigns to parameters as per the positional arguments

def sub(x:int,y:int)->int:
    c= x-y
    return c
# function calling block
a,b=10,20
res = sub(a,b)
print(res)
print(sub(20,10))
```

```python
#default arguments: the parameters defined with some default values,and if you pass the arguments for default parameter 
             otherwise, it will assign default values
def mul(a,b,c=0,d=0):
    res=a*b*c*d
    return res
print(mul(10,20))
print(mul(10,20,5))
print(mul(10,20,2,1))
```
