```python
#Element adding methods
- append(item) :it adds item at end of the list
- extend(iterable) :it adds all iterable elements at end of the list
- insert(pos,item) :this method adds ther item at specified position

```

```python
lst=[1,2,3,4]
lst.append(10)
print(lst)
lst.append('code')
print(lst)
lst.append([10,20,30])
print(lst)
print(lst[6])
lst.extend([100,200,300])
print(lst)
lst.insert(1,2000)
print(lst)
```

```python
#element removing methods
 -remove(val):
-pop(): it removes the first occurance of the value if value exists in list,otherwise it raises 'ValueError'
-pop(pos):it removes and returns the specified position elements from list 
-clear(): it removes all the elements from list
-del():it python keyword ,used to delete values.
```

```python
lst=[10,20,30,20,40]
lst.remove(20)
print(lst)
val=lst.pop()
print(val,lst)
val2=lst.pop(1)
print(val2,lst)
del lst[0]
print(lst)
lst.clear()
print(lst)
```
