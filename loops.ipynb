```python
#count  even and odd numbers given list of numbers
numbers=[10,20,30,27,7,8,13,20]
even_count =0
odd_count=0
for num in numbers:
    if num % 2==0:
        even_count+=1
    else:
        odd_count+=1
print("even numbers count:",even_count)
print("odd numbers count:",odd_count)
```

```python
#seperate even and odd numbers in given list of numbers
numbers=[10,15,22,7,8,13,20]
even=[]
odd=[]
for num in numbers:
    if num % 2==0:
        even.append(num)
    else:
        odd.append(num)
print("even numbers:",even)
print("odd numbers:",odd)
```
