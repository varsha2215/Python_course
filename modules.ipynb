```python
import random 
print(random.random())#this function returns the value in between 0 and 1
print(random.randint(1,10))#randint(a,b): this function generates random number between 
print(random.choice(['A',1,2,3,4,5,6,7,8,9,'J','Q','K']))
print(random.choices(['A',1,2,3,4,5,6,7,8,9,'J','Q','K'],k=3))
arr=['A',1,2,3,4,5,6,7,8,9,'J','Q','K']
random.shuffle(arr)
print(arr)
```

```python
#who will reach top in less steps
import random
steps=rolls=0
while True:
    input("To roll dice for that click enter button")
    dice_val=random.randint(1,6)
    rolls+=1
    print("current dice value:",dice_val)
    if steps + dice_val <=100:
        steps +=dice_val
    if steps==100:
        print(f"your reached TOP in {rolls} rolls")
        break
print(f"for {rolls} rolls your at {steps} position")
```
