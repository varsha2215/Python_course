```python
##with parameters and with return
#sum of two numbers
#function definition block
def add(x:int,y: int)->int:
    c=x+y
    return c
#function calling
a,b=5,6
res=add(a,b)
print(res)
print(add(1,2))
print(add(3,4))
```

```python
#with return and without parameter
#sum of two numbers
#function definition
def add()->int:
    x,y=10,20
    c=x+y
    return c
#function calling
res=add()
print(res)
print(add())
print(add())
print(add())
```
