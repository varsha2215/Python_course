### String Methods
-Case Conversion Methods
  -upper()
  -lower()
  -title()
  -swapcase()
  -captalize()
  

```python
String ="The ocean is blue and sky is BLUE "
lower=String.lower()
print(lower)
print(String.upper())
print(String.title())
print(String.swapcase())
print(String.capitalize())
```

```python
#white space remove methods
#strip()
#lstrip()
#rstrip()
string="   codegnan   "
print(string)
print(string.strip())
print(string.lstrip())
print(string.rstrip())
```

```python
#split():
string="The ocean is blue and sky is BLUE 123442"
print(string.split())
print(string.split('is'))
print(string.split('z'))
print(string.split('@'))
```

```python
#indexing finding methods
#index(sub_string)
#find(sub_string)
string="the ocaen is blue and sky is blue"
print(string.index("o"))
print(string.index("is"))
print(string.index('is',14))
```

```python

print(string.find("is"))
print(string.find('is',14))
print(string.find('z'))
```

```python
#replace (old,new)
string="the ocaen is blue and sky is blue"
print(string.replace('blue','red'))
print(string.replace('blue','red',1))

```

```python
#"sep".join(iterable):
lst=['the','ocean','is','blue','and','sky','is','BLUE']
print("".join(lst))
print("".join(lst))
print("_".join(lst))
```
