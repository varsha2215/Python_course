```python
#enclosing scope
def outer(x):
    y=10 #enclosing scope
    def inner():
        a=10 #local scope
        z=y+a
        print(f"This is local scope:{z}")
    inner()
    y=y+x
    print(f"This is enclosing scope:{y}")
n=5
outer(n)
        
```

```python
# global scope
x=10 #Global scope
def outer():
    y=20 #Enclosing scope
    def inner():
        z =30 #local scope
        res=z+y+x
        print(f"This is local scope:{res}")
    y=y+x
    inner()
    print(f"This is enclosing scope:{res}")
outer()
```

```python
#update the global scope variable inside the local scope by using 'globval' keyword
x=10 # Global scope
def outer():
    global x
    x = x+1
    print(f"This is Local scope val:",x)
outer()
print(f"This is Global scope val:",x)

        
```

```python
def outer_function():
    # Enclosing scope variable
    count = 10  

    def inner_function():
        # Declare count as nonlocal to update the enclosing scope
        nonlocal count  
        count = 20  # Updates the outer variable

    inner_function()
    print(count)  # Outputs 20

outer_function()
```
