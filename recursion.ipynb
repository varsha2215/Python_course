```python
def sum_n(n):
    # Base condition
    if n == 0:
        return 0

    # Recursive condition
    return n + sum_n(n - 1)

print(sum_n(5))

```

```python
#find binary number for decimal number using recursion
def decimal_to_binary(n):
    if n == 0:
        return

    decimal_to_binary(n // 2)  # Recursive call
    print(n % 2, end="")

num = 5
decimal_to_binary(num)
```

```python
def fibinocci(n):
    if n==1 or n==0:
        return 1
    else:
        return fibinocci(n-1)+fibinocci(n-1)
fibinocci(6)
```

```python
def binary(n):
    if n==0:
        return 0
    if n==1:
        return 1
    return str (binary(n//2))+str(n%2)
binary(15)
```

```python
def naturalnum(n):
    if n==0:
        return 1
    else:
        return n+naturalnum(n-1)
print(naturalnum(6))
```

```python
#find square of all list of integers using lambda and map()
arr=[1,2,3,4,5]
res=list(map(lambda x : x**2,arr))
print(res)

```

```python

#find square of all list of integers using lambda and filter
arr=[1,2,3,4,5]
res=list(filter(lambda x : x%2==0,arr))
print(res)


```
